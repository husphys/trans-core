# xLSTM depth-checkpoint transfer audit

**Purpose:** Verify all five MagNet xLSTM depth checkpoints transfer only the waveform encoder and xLSTM backbone.

**Scientific status:** AUDIT ONLY; no scientific fine-tuning.

**Prerequisites:** Completed Notebook 21 depth artifacts for depths 2, 4, 6, 8, and 10.

**Dataset/evidence paths:** experiments/xlstm_depth_v1/depth_{2,4,6,8,10}/best_checkpoint.pt


```python
from pathlib import Path
import sys
import os
PROJECT_ROOT = Path.cwd().resolve()
FORCE_RETRAIN = False
AUTO_RESUME = True
REQUIRED_INTERPRETER = os.environ.get('MEPI_REQUIRED_INTERPRETER')
print(f'sys.executable: {sys.executable}')
print('Required kernel: Python (trans-core)')
print(f'FORCE_RETRAIN={FORCE_RETRAIN}; AUTO_RESUME={AUTO_RESUME}')
if REQUIRED_INTERPRETER and str(Path(sys.executable).resolve()) != str(Path(REQUIRED_INTERPRETER).resolve()):
    raise RuntimeError('Select Python (trans-core), Restart Kernel, then Run All')
print('ENVIRONMENT CHECK: PASS')
```

```python
from src.mepi_v1.checkpointing import audit_xlstm_transfer_checkpoint
DEPTHS = (2, 4, 6, 8, 10)
reports = []
for depth in DEPTHS:
    try:
        report = audit_xlstm_transfer_checkpoint(PROJECT_ROOT, depth)
        reports.append(report)
        print(f'Depth {depth} checkpoint: {report["checkpoint_path"]}')
        print(f'SHA256: {report["checkpoint_sha256"]}')
        print('Waveform encoder keys:', report['waveform_encoder_keys'])
        print('xLSTM backbone keys:', report['backbone_keys'])
        print('Temporary material-head keys (discarded):', report['temporary_material_head_keys'])
        print('MagNet 2-feature encoder keys (discarded):', report['magnet_operating_encoder_keys'])
        print('Transferred tensor shapes:', report['transferred_shapes'])
        print('Discarded tensor shapes:', report['discarded_shapes'])
        print(f'TRANSFER AUDIT: PASS — depth {depth}')
    except Exception as error:
        print(f'TRANSFER AUDIT: FAIL — depth {depth}: {error}')
        raise
if len(reports) != 5:
    raise RuntimeError('Transfer audit did not pass for all five depths')
print('5/5 DEPTH CHECKPOINTS TRANSFER-READY')
```
