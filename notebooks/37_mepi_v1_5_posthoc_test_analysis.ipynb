# Notebook 37 — MEPI v1.5 post-hoc final-test analysis

This notebook reports additional metrics and sample-level predictions for the already frozen final model. It does not train, select, tune, or change a checkpoint. The exactly-once final-test ledger is verified and left unchanged.


```python
from pathlib import Path
import sys
import torch

POSTHOC_ANALYSIS_ONLY = True
TRAINING_PERFORMED = False
MODEL_SELECTION_PERFORMED = False
HYPERPARAMETER_CHANGED = False
CHECKPOINT_CHANGED = False
RUN_POSTHOC = True
PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / 'MEPI-FROZEN-PROTOCOL v1.5.md').is_file():
    raise RuntimeError('Run from the repository root or notebooks directory')
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
assert all((POSTHOC_ANALYSIS_ONLY, not TRAINING_PERFORMED, not MODEL_SELECTION_PERFORMED, not HYPERPARAMETER_CHANGED, not CHECKPOINT_CHANGED))
print('POSTHOC_ANALYSIS_ONLY = TRUE')
print('TRAINING_PERFORMED = FALSE')
print('MODEL_SELECTION_PERFORMED = FALSE')
print('HYPERPARAMETER_CHANGED = FALSE')
print('CHECKPOINT_CHANGED = FALSE')
```

```python
from src.mepi_v1.posthoc_analysis_v1_5 import ERR95_EPSILON, ERR95_SOURCE, run_posthoc_analysis

assert ERR95_EPSILON == 1e-12
assert ERR95_SOURCE == 'src/evaluation/metrics.py'
device = 'cuda' if torch.cuda.is_available() else 'cpu'
if not RUN_POSTHOC:
    raise RuntimeError('Post-hoc report generation was not authorized in this notebook')
result = run_posthoc_analysis(PROJECT_ROOT, device=device)
assert result['POSTHOC_ANALYSIS_ONLY'] is True
assert result['TRAINING_PERFORMED'] is False
assert result['MODEL_SELECTION_PERFORMED'] is False
assert result['HYPERPARAMETER_CHANGED'] is False
assert result['CHECKPOINT_CHANGED'] is False
assert result['FINAL_TEST_LEDGER_MODIFIED'] is False
assert all(item['within_1e-5'] for item in result['agreement_with_frozen_final_report'].values())
print('POSTHOC_REPORTING_COMPLETE = TRUE')
print('DEVICE =', device)
print('ERR95_EPSILON =', ERR95_EPSILON)
print('CHECKPOINT_SHA256 =', result['checkpoint_sha256'])
```

## Interpretation boundary

LSP is a dimensionless Arrhenius-informed relative thermal-stress proxy, not lifetime, RUL, time-to-failure, or measured degradation. Predicted LSP standard deviation is not described as calibrated uncertainty. Err95 is a dataset-level test metric, not a per-sample confidence measure.
